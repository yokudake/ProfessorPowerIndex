```python
import pandas as pd
import numpy as np

# Plotting
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.feature_selection import SelectKBest, mutual_info_classif

# Misc tools
from tqdm.notebook import tqdm
import os
import re
import time
import warnings
warnings.filterwarnings("ignore")
```

```python
model_df = pd.read_csv('2023-03-28_NB4_ml_df.pkl')
```

```python
X_train = model_df.drop('PaperTier')
y_train = model_df['PaperTier']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_train, y_train, test_size=0.25, random_state=1)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# 1. Instantiate model
logit = LogisticRegression(solver='lbfgs', random_state=1)

# 2. Fit model
logit.fit(X_train, y_train);

# 3. Score model
print('Score on train: {:0.2%}'.format(logit.score(X_train, y_train)))
print('Score on test: {:0.2%}'.format(logit.score(X_test, y_test)))
```
