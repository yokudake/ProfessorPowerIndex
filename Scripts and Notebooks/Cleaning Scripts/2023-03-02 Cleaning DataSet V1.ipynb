```python
df3 = df2.drop(['StartPage', 'EndPage', 'Age'], axis = 1)
```

```python
num = df3.isna().sum().sum()
print(f"There are {num} null values")
```

```python
df4 = df3.dropna(subset=['Authors', 'Year'])
```

```python
df4.isna().sum()
```

```python
df4.head()
```

```python
rows, cols = df4.shape
print(f'There are {rows} rows')
```

```python
df4['Source'].value_counts()
```

```python
import re
pattern = r'^(.*?)\s+(\d+)\s+\((\d+)\),\s+(.*)$'

vals = df4['Source'].values
index = df4['Source'].index

for idx, val in enumerate(df4['Source'].values):

    # match pattern to citation
    match = re.match(pattern, val)

    if match:
        journal = match.group(1)   # 'Optics Express'
        volume = match.group(2)    # '27'
        issue = match.group(3)     # '20'
        pages = match.group(4)     # 'A1419-A'

        # print extracted values
        print(val)
        print('Journal:', journal)
        print('\n')
        
        df4.loc[index[idx], 'Source'] = journal
        
    else:
        pass
```

```python
df4['Source'].value_counts()
```

```python
df5 = df4.assign(Source = df4['Source'].apply(lambda row: np.char.capitalize(str(row))))
```

```python
df5['Source'].value_counts()
```

```python
import re
pattern = r'^([^,]+)'

vals = df5['Source'].values
index = df5['Source'].index

for idx, val in enumerate(df5['Source'].values):

    # match pattern to citation
    match = re.match(pattern, val)

    if match:
        journal = match.group(1)
        
        # print extracted values
        print(val)
        print('Journal:', journal)
        print('\n')
        
        df5.loc[index[idx], 'Source'] = journal
        
    else:
        pass
```
