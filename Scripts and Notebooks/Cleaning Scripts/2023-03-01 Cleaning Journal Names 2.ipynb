Let's get an array of all the unique journal names.

```python
jour_ser = df['journal']
print(f"There are {len(jour_ser.unique())} of unique names")
```

```python
for i in jour_ser.unique():
    print(i)
```

First, let's replace all the instances with `arXiv` as just `arXiv`

```python
replace_dic = {}
for i in jour_ser.unique():
    if 'arXiv' in i:
        replace_dic.update({i: 'arXiv'})

df1 = df.replace(replace_dic)
```

```python
jour_ser = df1['journal']
print(f"There are {len(jour_ser.unique())} of unique names")
```

```python
for i in jour_ser.unique():
    print(i)
```
