Let's test an algorithm for replacing variations in names using the `level_0_author` = `Edward Sargent`

```python
sarg_df = df[df['level_0_author'] == 'Edward Sargent']
```

```python
sarg_df.head()
```

We want to find all the variations of the name *Edward Sargent* in the co-author columns and replace them with one consistent name. Let's start with creating a series of all the unique co-author names.

```python
cols_to_exclude = sarg_df.loc[:, 'total_citations':'publication_year'].columns
cols_to_keep = sarg_df.columns.difference(cols_to_exclude)
result_df = sarg_df.loc[:, cols_to_keep]
```

```python
def pull_names(row):
    # Returns the values of the authors in each row
    return row.values

grouped_df = result_df.groupby('level_0_author')

for name, group in grouped_df:
    result = [name for name_list in group.values for name in name_list if isinstance(name, str)] #make sure the level_0_author name gets added
    result.append(name)
    
name_ser = pd.Series(result)
```

We have our series `name_ser` let's see the distribution

```python
name_ser.value_counts(normalize = True)
```

Now let's find the names that are most closely related to the of `level_0_author` names using the package `SequenceMatcher`.

```python
dic = {}
key = 'Edward Sargent'
dic[key] = {}

names = name_ser.value_counts().index
name_cts = name_ser.value_counts().values

for idx, name in enumerate(names):
    ratio = SequenceMatcher(None, key, name).ratio()
    if ratio > 0.75:
        dic[key].update({name: name_cts[idx]})
```

We now have a dictionary of all the names that match the `level_0_author` value by at least $75\%$

```python
dic
```

Let's replace all the other names with the name that occurs the most.

```python
max_cts = np.max([dic[key][i] for i in dic[key].keys()])
max_key, = [i for i in dic[key].keys() if dic[key][i] == max_cts]

print(f"The name that occurs the most is {max_key} with {max_cts} occurances")
```

First, create a dictionary to to use the `.replace()` method

```python
replace_dic = {}
key = max_key

names = dic[key].keys()
for idx, name in enumerate(names):
    replace_dic[name] = key
```

```python
replace_dic
```

This dictionary will replace all the variations with `Edward Sargent`

```python
sarg_df = sarg_df.replace(replace_dic)
```

```python
sarg_df.head()
```

The length of the `name_ser` went down from 2355 to 2350, representing the 5 other versions of *Edward Sargent* that were overwritten.

```python
cols_to_exclude = df.loc[:, 'total_citations':'publication_year'].columns
cols_to_keep = df.columns.difference(cols_to_exclude)
result_df = df.loc[:, cols_to_keep]

def pull_names(row):
    # Returns the values of the authors in each row
    return row.values

replace_dic = {}
grouped_df = result_df.groupby('level_0_author')

for name, group in grouped_df:
    result = [name for name_list in group.values for name in name_list if isinstance(name, str)] #make sure the level_0_author name gets added
    result.append(name)
    name_ser = pd.Series(result)
    
    dic = {}
    key = name
    dic[key] = {}

    names = name_ser.value_counts().index
    name_cts = name_ser.value_counts().values

    for idx, name in enumerate(names):
        ratio = SequenceMatcher(None, key, name).ratio()
        if ratio > 0.75:
            dic[key].update({name: name_cts[idx]})
    
    max_cts = np.max([dic[key][i] for i in dic[key].keys()])
    max_key, = [i for i in dic[key].keys() if dic[key][i] == max_cts]

    print(f"The name that occurs the most is {max_key} with {max_cts} occurances")
    
    import re
    
    # remove any non-UTF-8 characters using encode method
    new_key = max_key.encode('ascii', 'ignore').decode('ascii')
    
    # Use re to remove . inside parentheses using regex
    new_key = re.sub(r'\(.\)', '', new_key).strip()
    
     # Use re to remove empty parentheses using regex
    new_key = re.sub(r'\(\)', '', new_key).strip()
    
    # Use re to remove Dr inside parentheses using regex
    new_key = re.sub(r'\(Dr\)', '', new_key).strip()
    
    # Use re to remove title: , Ph.D
    new_key = re.sub(r',\s*\(?Ph\.?D.?\)?.*$','', new_key).strip()
    
    # Use re to remove title: Dr
    new_key = re.sub(r'^Dr\.?\s*', '', new_key).strip()
    
    # Use re to remove title: Proff.
    new_key = re.sub(r'^PROFF\.?\s*', '', new_key).strip()
    
    # toggling the case for consistency, remove non-ASCII characters
    new_key = str(np.char.capitalize(new_key))

    print("New key is ", new_key)
    names = dic[max_key].keys()
    for idx, name in enumerate(names):
        replace_dic[name] = new_key 
```

Let's now replace the old names with the new ones with the `replace_dic` dictionary

```python
new_df = df.replace(replace_dic)
```

Doing some additional cleaning. Removing lines where there is a co-author name with less than 3 characters.

```python
new_df = pd.read_csv(r"G:\My Drive\BrainStation\Capstone\Data\2023-02-28 Name Cleaning 1_V1.csv", low_memory = False)
```

```python
row_before = new_df.shape[0]
new_df.info()
```

```python
def pull_names(row):
    # Returns the values of the authors in each row
    i = 0 
    while i < len(row.values):
        val = row.values[i]
        if isinstance(val, str):
            if len(val) < 3:
                return True # if a name is found, return True
                break # breaks the loop if a author with less than 3 characters is found name is found
        i = i + 1
    return False # if no names are found, return False

cols_to_exclude = new_df.loc[:, 'total_citations':'publication_year'].columns
cols_to_keep = new_df.columns.difference(cols_to_exclude)
result_df = new_df.loc[:, cols_to_keep]

rows = result_df.apply(pull_names, axis = 1)
new_df = new_df.drop(rows[rows].index, axis = 0)
```

```python
row_after = new_df.shape[0]
diff = row_before - row_after
print(f"{diff} rows were removed")
```

Rewriting to a new `.csv` file
