```python
def rest_ttsplit(ts):
    X_train = model_df.drop('PubTier', axis = 1)
    y_train = model_df['PubTier']
    return train_test_split(X_train, y_train, test_size=ts, random_state=1)
```

```python
test_sizes = np.arange(0.025, 0.925, 0.025)
train_scores = []
test_scores = []

TrainTestPred = {'train': [], 'test':[], 'pred':[], 'X_test':[]}

for ts in tqdm(test_sizes): 
    
    # Reset the Train Test Split
    X_train, X_test, y_train, y_test = rest_ttsplit(ts)
    
    # Score the model
    scaler = StandardScaler()
    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)
    
    # 1. Instantiate model
    logit = LogisticRegression(solver='lbfgs', random_state=1)

    # 2. Fit model
    logit.fit(X_train, y_train);
    
    # 3. Append Scores
    train_scores.append(logit.score(X_train, y_train))
    test_scores.append(logit.score(X_test, y_test))
    
    # Get class predictions
    y_pred = logit.predict(X_test)
    
    TrainTestPred['X_test'].append(X_test)
    TrainTestPred['train'].append(y_train)
    TrainTestPred['test'].append(y_test)
    TrainTestPred['pred'].append(y_pred)
    
```

```python
plt.plot(test_sizes, train_scores, marker = 'o', label = "Train Score")
plt.plot(test_sizes, test_scores, marker = 'o', label = "Test Score")
plt.xlabel('Size of the Test Set (% of Total Observations)', fontsize = 14, labelpad = 10)
plt.grid()

x_ticks = np.arange(0, 1, 0.1)
x_labels = ['{:.0%}'.format(i) for i in x_ticks]
plt.xticks(x_ticks, x_labels);

y_ticks = np.arange(0.985, 1.005, 0.005)
y_labels = ['{:.1%}'.format(i) for i in y_ticks]
plt.yticks(y_ticks, y_labels);

plt.legend()
```

As the percentage of observations in the test set increases, the less over-fitting begins to occur. Over-fitting would be best addressed using penalties, so optimizing on those parameters should be the next step in optimizing the model.

Just try a pipeline

```python
X_train = model_author.drop('PubTier', axis = 1)
y_train = model_author['PubTier']
X_train, X_test, y_train, y_test =  train_test_split(X_train, y_train, test_size=0.10, random_state = 1)

#Define estimators
estimators = [
    ('scaler', StandardScaler()),
    ('model', DecisionTreeClassifier())
]

# Initiate Pipeline object
pipe = Pipeline(estimators)

# Define parameter grid for grid search
param_grid = {'model__min_samples_leaf': np.arange(1, 20, 1),
    'model__max_depth': np.arange(1, 50, 1),
}

# Initiate the GridSearchCV object with a 5-fold cross validation
grid = GridSearchCV(pipe, param_grid=param_grid, cv = 5)

# Perform the grid search, fitting the grid search to the training data only
grid.fit(X_train, y_train)

# Evaluate the performance on the validation set
val_score = grid.score(X_test, y_test)

# Print the best parameters, best score, and validation score
print(f"Best d parameters: {grid.best_params_}")
print(f"Best training score: {grid.best_score_}")
print(f"Validation score: {val_score}")
```
