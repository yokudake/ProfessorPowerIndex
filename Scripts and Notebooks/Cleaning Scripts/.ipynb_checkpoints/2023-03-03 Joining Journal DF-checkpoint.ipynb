Start with `energy &environmental science `

```python
def find_word(row, word):
    result = np.char.find(row['Title'], word)
    if result != -1:
        print(result, row['Title'])
        return True
    elif result > -1:
        return False
```

```python
jour_df.apply(find_word, word = 'energy and', axis = 1);
```

Replace with `energy and environmental science`

```python
def genjoin(replace_dic):
    
    global df 
    
    # replace the words in the original data frame
    df = df.replace(replace_dic)
    
    # Perform Left Join
    left_join = pd.merge(left = df, right = jour_df, left_on = 'Source', right_on = 'Title', how = 'left')
    
    # Find Null Values
    nan_idx = left_join['id'][left_join['id'].isna() == True].index
    nan_df = left_join.loc[nan_idx, :]
    
    print(f"There are {left_join.isna().sum().sum()} Remaining Nulls")

    return nan_df['Source'].value_counts()
```

```python
genjoin({'energy &environmental science':'energy and environmental science'})
```
