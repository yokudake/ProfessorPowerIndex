Some of the values in `article_title` could actually have part of the citation in them. Let's see if we can find them. We can do that by using the `re` package to search for the format of an ACS citation.

```python
test_str = 'Energy Environ. Sci. 9, 1989 (2016)'
```

```python
test_str2 = 'Perovskite light-emitting diodes with external quantum efficiency exceeding 20 per cent'
```

```python
test_str3 = 'High-quality bulk hybrid perovskite single crystals within minutes by inverse temperature crystallization. Nat. Commun. 6, 7586 (2015)'
```

```python
import re

# For journal, year
#pattern = r'^(.+?)\s(\d+),\s(\d+)\s\((\d{4})\)'

pattern = r'^(.+?)\.\s((?:\w+\W+){1,3}?)\s(\d+),\s(\d+)\s\((\d{4})\)'
for s in [test_str, test_str2, test_str3]:
    #print(s)
    match = re.match(pattern, s)
    if match != None:
        title = match.group(1)
        journal_name = match.group(2)
        volume_number = match.group(3)
        page_number = match.group(4)
        year = match.group(5)
        print(title, journal_name, year)
    else:
        print('No Match')
```

This seems to work well for ACS citation, let's run it.

```python
idx = df['article_title'][rows].index
vals = df['article_title'][rows].values

pattern = r'^(.+?)\.\s((?:\w+\W+){1,3}?)\s(\d+),\s(\d+)\s\((\d{4})\)'
for i, s in enumerate(vals):
    #print(s)
    match = re.match(pattern, s)
    if match != None:
        title = match.group(1)
        journal_name = match.group(2)
        volume_number = match.group(3)
        page_number = match.group(4)
        year = match.group(5)
        print(idx[i])
        print(title, journal_name, year)
    else:
        pass
```

```python
idx = 1578
vals = df['article_title'][1578]

print(df.loc[idx, :].values)

pattern = r'^(.+?)\.\s((?:\w+\W+){1,3}?)\s(\d+),\s(\d+)\s\((\d{4})\)'

match = re.match(pattern, vals)
if match != None:
    title = match.group(1)
    journal_name = match.group(2)
    volume_number = match.group(3)
    page_number = match.group(4)
    year = match.group(5)
    if title != None:
        df.loc[idx, 'article_title'] = title
    if journal_name != None:
        df.loc[idx, 'journal'] = journal_name
    if year != None:
        df.loc[idx, 'publication_year'] = year
    print(title, journal_name, year)


print(df.loc[1578, :].values)
```

```python
# Tesst
test_str = 'Reduction of Hysteresis in Hybrid Perovskite Transistors by Solvent-Controlled Growth. Materials 2021, 14, 2573'
test_str1 = 'Electrical and Optical Characterization of CsPbCl3 Films around the High-Temperature Phase Transitions. Nanomaterials 2022, 12, 570'

pattern = r'^(.*?)\.\s*(.*?)\s*(\d+),\s*(\d+),\s*(.*?)$'

match = re.search(pattern, test_str)

if match:
    title = match.group(1)
    journal = match.group(2)
    year = match.group(3)
    volume = match.group(4)
    page = match.group(5)
    
    print('Title:', title)
    print('Journal:', journal)
    print('Volume:', volume)
    print('Year:', year)
```
