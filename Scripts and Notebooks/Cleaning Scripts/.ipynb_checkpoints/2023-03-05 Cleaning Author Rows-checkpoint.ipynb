Step 2 to cleaning names. For papers with repeating titles, find instances where the author names don't match and then change the author names so that they all match.

```python
def pull_name(row):
    '''
    For each title in the data set, check if there is more than one citation. If there is, check
    if the first authors names are formatted in the same way. If not, print those names.
    '''
    a1 = row['first_auth']
    if len(a1) > 1:
        if a1.values[0] != a1.values[1]:
            print(row['Title'])
            print(a1)


df.groupby('Title').apply(pull_name)
```

For each one of the groups of authors, check to make sure that the last names are the same. If they are, use the `auth_val` series to find which name appears more often and replace the other names with the most occurring name.

```python
def all_same(itemlist):
    '''
    Dermines if all the elements in a list are the same.
    
    '''
    return all(x == itemlist[0] for x in itemlist)

def pull_name(row, auth_type = ''):
    '''
    For each title in the data set, check if there is more than one citation. If there is, check
    if the first authors names are formatted in the same way. If not, print the variation of the 
    name that is the longest.
    '''
    replace_dic = {}
    a1 = row[auth_type]
    if len(a1) > 1:
        if a1.values[0] != a1.values[1]:
            last_names = [i.rsplit(' ')[-1] for i in a1.values if isinstance(i, str)] # make a list of the last names
            if all_same(last_names): # Check if the last names are all the same.
                top_length = 0
                top_name = ''
                for i in a1.values:
                    if isinstance(i, str):
                        name, cts = auth_val[auth_val.index == i].index, auth_val[auth_val.index == i].values
                        length = len(name[0])
                        if length > top_length:
                            top_length = length
                            top_name = name
                    else:
                        pass
                for i in a1.values:
                    if isinstance(i, str):
                        replace_dic[i] = top_name[0]
                    else:
                        pass
                
                return replace_dic
            
            else:
                return False
        else:
            return False
    else:
        return False
```

```python
replace_dic = {}
return_vals = df.groupby('Title').apply(pull_name, auth_type = 'first_auth').values
return_dics = return_vals[return_vals != False]

for dic in return_dics:
    replace_dic.update(dic)
    
replace_dic
```

It looks like for some of the names the last names got cut off or abbreviated. ('dequilettes' vs 'quilettes', for example)

```python
from difflib import SequenceMatcher

def pull_name2(row, auth_type = ''):
    '''
    For each title in the data set, check if there is more than one citation. If there is, check
    if the first authors names are formatted in the same way. If not, print the variation of the 
    name that is the longest.
    '''
    a1 = row[auth_type]
    if len(a1) > 1:
        if all_same(a1.values) == False: # if all the elements in the string aren't the same
            last_names = [i.rsplit(' ')[-1] for i in a1.values if isinstance(i, str)] # make a list of the last names
            if all_same(last_names) == False: #if all the last names aren't the same
                replace_dic = {}
                for i in range(0, len(a1.values)):
                    for j in range(i+1, len(a1.values)):
                        a = a1.values[i]
                        b = a1.values[j]
                        if a != b:
                            r = SequenceMatcher(None, a, b).ratio()
                            if r > 0.70: # if the names are at least 70% similar
                                name = [a,b]
                                name.sort(reverse = False, key = len)
                                replace_dic.update({name[0] : name[1]})
                
                if len(replace_dic) != 0: #if replace_dic is not empty
                    return replace_dic 
                else:
                    return False
            
            else:
                return False
    
        else:
            return False
    
    else:
        return False
        
            
```

```python
replace_dic = {}
return_vals = df.groupby('Title').apply(pull_name2, auth_type = 'first_auth').values
return_dics = return_vals[return_vals != False]

for dic in return_dics:
    replace_dic.update(dic)
    
replace_dic
```

```python
df = df.replace(replace_dic)
get_unique_auth()
```

About 3 author names got replaced. Now let's repeat this process but with grouping by `query_id` instead of `Title`. We can combine `pull_name` and `pull_name2` into one function to expedite the process.

```python
def pull_name3(row, auth_type = ''):
    '''
    For each title in the data set, check if there is more than one citation. If there is, check
    if the first authors names are formatted in the same way. If not, print the variation of the 
    name that is the longest.
    '''
    a1 = row[auth_type]
    if len(a1) > 1:
        replace_dic = {}
        if all_same(a1.values) == False:
            last_names = [i.rsplit(' ')[-1] for i in a1.values if isinstance(i, str)] # make a list of the last names
            if all_same(last_names): # Check if the last names are all the same.
                top_length = 0
                top_name = ''
                for i in a1.values:
                    if isinstance(i, str):
                        name, cts = auth_val[auth_val.index == i].index, auth_val[auth_val.index == i].values
                        length = len(name[0])
                        if length > top_length:
                            top_length = length
                            top_name = name
                    else:
                        pass
                for i in a1.values:
                    if isinstance(i, str):
                        replace_dic[i] = top_name[0]
                    else:
                        pass
            else: # if the last names aren't the same
                for i in range(0, len(a1.values)):
                    for j in range(i+1, len(a1.values)):
                        a = a1.values[i]
                        b = a1.values[j]
                        if a != b:
                            r = SequenceMatcher(None, a, b).ratio()
                            if r > 0.70: # if the names are at least 70% similar
                                name = [a,b]
                                name.sort(reverse = False, key = len)
                                replace_dic.update({name[0] : name[1]})
                
            if len(replace_dic) != 0: #if replace_dic is not empty
                return replace_dic 
        else:
            return False
    else:
        return False
```

```python
replace_dic = {}

for a in ['first_auth', 'second_auth', 'third_auth']:
    return_vals = df.groupby('query_id').apply(pull_name, auth_type = a).values
    return_dics = return_vals[return_vals != False]
    for dic in return_dics:
        replace_dic.update(dic)
    
replace_dic
```

```python
df = df.replace(replace_dic)
get_unique_auth()
```

It looks like only one name got replaced. Let's try to group the co-author names by last names and see if we can make any replacements.

```python
last_names = set([i.rsplit(' ')[-1] for i in auth_ser.unique() if isinstance(i, str)]) # make a list of the last names
```

```python
def get_names(word):
    '''
    
    Checks to see if a word is in the list of last names.
    
    Parameters:
    -----------
    word: a string
    
    Return:
    -------
    word_bank: a list of names that have the word in them
    
    '''
    word_bank = []
    for i in set(auth_ser.unique()):
        if isinstance(i, str): #checking to make sure not a null
            result = np.char.find(i, word)
            if result != -1:
                if SequenceMatcher(None, i, word).ratio() > 0.50:
                    word_bank.append(i)
    return word_bank
```

```python
from tqdm.notebook import tqdm

# This will take about 20-30 minutes...
groups = [get_names(i) for i in tqdm(last_names)]
groups
```

```python
new_groups = [i for i in groups if len(i) > 1]
```

```python
new_groups
```

```python
#group_df.to_csv('new_groups.csv')
```

If they start with the same letter, group together

```python
def match_first(names):
    match_first = {}
    first_names = [i.rsplit(' ')[0][0] for i in names if isinstance(i, str)] # make a list of the last names
    
    top_cts = 0
    top_name = ''
    name_replace = []
    
    for i in range(0, len(first_names)):
        for j in range(i+1, len(first_names)):
            a = first_names[i]
            b = first_names[j]
        
            if a == b:
                name = [names[i], names[j]]
                name.sort(reverse = True, key = len)
                name_replace.append(names[i])
                name_replace.append(names[j])
                if len(name[0]) > top_cts:
                    top_cts = len(name[0])
                    top_name = name[0]
            
        for i in name_replace:
            match_first.update({i : top_name})
    
    if len(match_first) != 0:
        return match_first
```

```python
groups2 = [i for i in new_groups if len(i) <= 5]

replace_dic = {}
groups3 = [replace_dic.update(match_first(i)) for i in tqdm(groups2) if match_first(i)!= None];
replace_dic
```

```python
df = df.replace(replace_dic)
get_unique_auth()
```
