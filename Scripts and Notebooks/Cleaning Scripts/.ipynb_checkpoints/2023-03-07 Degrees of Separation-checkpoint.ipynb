**Step 1** Generate a unique list of authors `auth_uni`

```python
author_list = np.concatenate(df.loc[:, ['first_auth', 'second_auth', 'third_auth','query_author']].values)
auth_uni = pd.Series(author_list).unique()
```

**Step 2** Generate The Adjunct Matrix

```python
from tqdm import tqdm

# Create the adj matrix
n = auth_uni.shape[0]
adj_matrix = np.zeros((n, n))

for i in tqdm(range(df.shape[0])):
    frame = df.loc[i, ['first_auth', 'second_auth', 'third_auth','query_author']]
    frame = frame[frame.isna() == False]
    
    h = frame.values
    for i in range(len(h)):
        # loops over the name of one person (First iteration, Max)
        for j in range(i+1, len(h)):
            row, col = np.where(auth_uni == h[i]), np.where(auth_uni == h[j])
            adj_matrix[row, col] = 1
            adj_matrix[col, row] = 1
```

```python
adj_matrix.shape
```

Using the following code to find the degrees of separation between each co-author and query

One optimization you could try is to use a matrix of zeros and ones instead of NaNs to represent the graph. You can set matrix[i, j] to 1 if people[i] and people[j] are connected by an edge, and 0 otherwise. Then you can use matrix multiplication to compute the shortest paths between all pairs of people. This approach is based on the Floyd-Warshall algorithm.

```python
# initialize the distance matrix
print('initialize the distance matrix')
dist_matrix = np.where(adj_matrix == 1, 1, np.inf)
np.fill_diagonal(dist_matrix, 0)
```

```python
def floyd_warshall(graph):
    dist = np.copy(graph)
    n = graph.shape[0]
    for k in tqdm(range(n)):
        dist = np.minimum(dist, np.add.outer(dist[:, k], dist[k, :]))
    return dist
```

```python
dist_matrix = floyd_warshall(dist_matrix)
```
