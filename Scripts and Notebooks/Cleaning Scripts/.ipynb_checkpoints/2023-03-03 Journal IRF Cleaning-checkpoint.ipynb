#### Additional Processing for Matching with Dataset

```python
import re

def edit_title(row):
    title = row
    
    # Get rid of 'The'
    title = re.sub(r'^The\s', '', title)
    
    # Get rid of colons
    title = re.sub(r':', '', title)
    
    # Get rid of hyphens
    title = re.sub(r'-', '', title)
    
    # Get rid of double spacing
    title = re.sub(r'  ', ' ', title)
    
    return title.lower()
    
df3 = df2_3.assign(Title = df2_3['Title'].apply(edit_title))
```

```python
def find_word(row, word):
    result = np.char.find(row['Title'], word)
    if result != -1:
        print(result, row['Title'])
        return True
    elif result > -1:
        return False

df3.apply(find_word, word = 'science', axis = 1);
```

```python
import re

def edit_title2(row):
    title = row
    
    # Get rid of 'The'
    title = re.sub(r'^The\s', '', title)
    
    # Get rid of colons
    title = re.sub(r':', '', title)
    
    # Get rid of hyphens
    title = re.sub(r'-', '', title)
    
    # Get rid of double spacing
    title = re.sub(r'  ', ' ', title)
    
    # Get rid of '(discontinued)'
    title = re.sub(r'\(discontinued\)', '', title)
    
    return title.strip().lower()
```
